```python
from openai import OpenAI

client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
messages = [
    {
        "role": "system",
        "content": """
        You are a friendly conversational assistant.
        Remember the context of our conversation.
        """
    }
]
```

```python
while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Conversation ended.")
        break

    # Add user's message to history
    messages.append({
        "role": "user",
        "content": user_input
    })

    # Send entire conversation
    response = client.chat.completions.create(
        model="openai/gpt-5.5",
        messages=messages
    )

    # Get AI response
    assistant_message = response.choices[0].message.content

    print("AI:", assistant_message)

    # Add AI response to history
    messages.append({
        "role": "assistant",
        "content": assistant_message
    })
```

```python
from openai import OpenAI

client = OpenAI()
```

```python
response = client.responses.create(
    model="gpt-5.5",
    instructions="""
    You are a friendly conversational assistant.
    Remember the context of our conversation.
    """,
    input="Hi! My name is Sita and I'm learning Python."
)

print("AI:", response.output_text)
```

```python
previous_response_id = response.id

print("Response ID:", previous_response_id)
```

```python
response = client.responses.create(
    model="gpt-5.5",
    previous_response_id=previous_response_id,
    input="What am I learning?"
)

print("AI:", response.output_text)
```

```python
previous_response_id = None

while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Conversation ended.")
        break

    if previous_response_id is None:

        response = client.responses.create(
            model="gpt-5.5",
            instructions="""
            You are a friendly conversational assistant.
            Keep track of the conversation context.
            """,
            input=user_input
        )

    else:

        response = client.responses.create(
            model="gpt-5.5",
            previous_response_id=previous_response_id,
            input=user_input
        )

    print("AI:", response.output_text)

    previous_response_id = response.id
```
