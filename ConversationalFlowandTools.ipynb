```python
sys_prompt = """
You are Python instruction. Answer only Python related questions. Be polite and helpful.
"""
```

```python
input_prompt = [
                                                                  
    {'role':'user', 
     'content':"What is inflation?"},
                                                                 
    {'role':'assistant', 
     'content':"Sorry, I cannot answer questions out of Python"},
                                                                 
    {'role':'user', 
     'content':'Explain classes'}]
```

```python
response_roll_based = openrouter_client.responses.create(model="minimax/minimax-m3:free",

                                                         input = input_prompt,

                                                         instructions = sys_prompt,

                                                         reasoning = {"effort":"low"},

                                                         max_output_tokens = 300
                                                        )
print(response_roll_based.output_text)
```

```python
# flow

# chat history

# input
messages = []
print("Write a prompt to initiate chat.")
while True:
    user_input = input("If you want to end chat write 'quit'.")
    messages.append({"role":"user", "content":user_input})
    
    if user_input == "quit":
        break
    response = openrouter_client.responses.create(model='minimax/minimax-m3:free',
                                                  input = messages,
                                                  instructions="Always ask if user wants to ask something else. Keep your answer short.")
    
    messages.append({"role":"assistant", "content":response.output_text})
    print(response.output_text)
```
