#### 7. Z-test

```python
# population (hypothetical mean)
pop_mean = 300_000_000

# population standard deviation
pop_std = 400

# confidence level and error rate (p-val threshold)
alpha = 0.05

# sample size (here size of our column)
sample_size = Acc_grs.shape[0]
sample_mean = Acc_grs.mean()

# z-score (sample mean - population mean) / square root of sample size
z_score = (sample_mean - pop_mean) / np.sqrt(sample_size)

# a) two tailed test  ---> sample mean == 300_000_000
# p-value (with stats.norm.cdf(z_score)) for each test type
p_value = 2 * stats.norm.cdf(abs(z_score))

# decision for each test type (with if else)
if p_value < alpha:
      print("Reject the null hypothesis")
else:
      print("Fail to reject the null hypothesis")

# b) right tailed test ---> sample mean > 300_000_000
# p-value (with stats.norm.cdf(z_score)) for each test type
p_value = stats.norm.cdf(z_score)


# decision for each test type (with if else)
if p_value < alpha:
      print("Reject the null hypothesis")
else:
      print("Fail to reject the null hypothesis")


# c) left tailed test ---> sample mean < 300_000_000
# p-value (with stats.norm.cdf(z_score)) for each test type
p_value = 1 - stats.norm.cdf(z_score)

# decision for each test type (with if else)

if p_value < alpha:
      print("Reject the null hypothesis")
else:
      print("Fail to reject the null hypothesis")
```

#### 9. Create a Class

```python
# Create a class named "Book"

# attributes: title, author, number_of_pages

# methods:

# --- get_info() ---> prints out title (str), author (str) and number of pages (int)

# --- is_long() ---> if number of pages is more than 200, returns True, else, returns False

class Book:
  def __init__(self, title, author, number_of_pages):
    self.title = title
    self.author = author
    self.number_of_pages = number_of_pages

  def get_info(self):
        return (f'Title is {self.title}, author is {self.author}, number of pages are {self.number_of_pages}')

  def is_long(self):
        return self.number_of_pages >= 200


book1 = Book(
    title="Murder on the Orient Express",
    author="Agatha Christie",
    number_of_pages=256
)

print(book1.get_info())
print(book1.is_long())
```

#### 10. Create a Child CLass

```python
# Create a child class from 'Book' named 'EBook' (electronic book)

# extra attributes: file_size (float)

# extra methods: 
# --- download(as) ---> as takes file format from (pdf, epub, doc) options, prints f"dowloaded as {as}"
# --- get_file_info() ---> return file size and book name (title)

class eBook(Book):
    def __init__(self, title, author, number_of_pages, file_size:float):
        super().__init__(title, author, number_of_pages)
        self.file_size = file_size

    def download(self, format_as):
        assert format_as in ["pdf", "epub", "doc"], "Download as (pdf, epub, doc)."
        print(f"The book {self.title}, has been downloaded as {format_as}")

    def get_file_info(self):
        return f'file size is {self.file_size}, and book name is {self.title}'

e_book1 = eBook(
    title="Murder on the Orient Express",
    author="Agatha Christie",
    number_of_pages=256,
    file_size = 20
)
e_book1.download("doc")
e_book1.get_file_info()
```

#### 12. Create a function with Error Handling

```python
# create a function that calculates estimated salary

# the base amount is 1000 AZN

# if a person is 'intern', he will get half of base amount
# if a person is 'junior', he will get base amount
# if a person is 'middle', he will get twice of base amount
# if a person is 'senior', he will get triple of base amount


# function has parameter named 'level' and it indicates seniority level


# IMPORTANT NOTE!!! User can put values only in the list ['intern', 'junior', 'middle', 'senior'], 
# if we put something else, it will raise error
```

```python
# Təxmini maaşı hesablayan funksiya yaradın.

# Baza məbləği 1000 AZN-dir.

# Əgər şəxs 'intern'-dirsə, baza məbləğinin yarısını alır.
# Əgər şəxs 'junior'-dursa, baza məbləğini alır.
# Əgər şəxs 'middle'-dirsə, baza məbləğinin iki qatını alır.
# Əgər şəxs 'senior'-dursa, baza məbləğinin üç qatını alır.


# Funksiya 'level' adlı parametr qəbul edir. Bu parametr işçinin seniority səviyyəsini göstərir.


# VACİB QEYD:
# İstifadəçi yalnız bu dəyərlərdən birini daxil edə bilər: ['intern', 'junior', 'middle', 'senior']
# Əgər başqa dəyər daxil edilərsə, xəta (error) qaytarılmalıdır.
```

```python
def calc_salary(level):
    base_amount = 1000
    assert level in ['intern', 'junior', 'middle', 'senior'], "Level must be in the list ['intern', 'junior', 'middle', 'senior']"

    if level == "intern":
        return base_amount/2
    elif level == "junior":
        return base_amount
    elif level == "middle":
        return base_amount * 2
    else:
        return base_amount * 3

calc_salary("senior")
```
